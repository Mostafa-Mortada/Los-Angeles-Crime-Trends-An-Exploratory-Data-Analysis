```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("crimes.csv")
```

```python
df['Vict Sex'] = df['Vict Sex'].fillna('Unknown')
df['Vict Descent'] = df['Vict Descent'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
print(df['Vict Sex'].value_counts())
```

```python
df['Vict Sex'] = df['Vict Sex'].replace({'X': 'Not prefere to say', 'H': 'Not prefere to say'})
```

```python
print(df['Vict Sex'].value_counts())
```

```python
df['Date Rptd'] = pd.to_datetime(df['Date Rptd'], errors='coerce')
df['DATE OCC'] = pd.to_datetime(df['DATE OCC'], errors='coerce')
```

```python
df['TIME OCC'] = df['TIME OCC'].astype(str).str.zfill(4)
df['TIME OCC'] = pd.to_datetime(df['TIME OCC'], format='%H%M').dt.time
```

```python
df.head()
```

```python
df['month'] = df['DATE OCC'].dt.month
```

```python
df['year'] = df['DATE OCC'].dt.year
```

```python
df['year'].value_counts().sort_index().plot(kind='bar')
plt.title('Crime Distribution by Year')
plt.xlabel('Year')
plt.ylabel('Number of Crimes')
plt.show()
```

```python
df['month'].value_counts().sort_index().plot(kind='bar')
plt.title('Crime Distribution by Month')
plt.xlabel('Month')
plt.ylabel('Number of Crimes')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(y='Vict Descent', data=df, order=df['Vict Descent'].value_counts().index)

plt.title('Victim Descent Distribution')
plt.show()
```

```python
color_map = {
    'M': 'cyan',
    'F': 'pink',
    'Unknown': 'yellow',
    'Not prefere to say': 'yellow',}
```

```python
gender_counts = df['Vict Sex'].value_counts()
colors = [color_map.get(gender, 'gray') for gender in gender_counts.index]
gender_counts.plot(kind='bar', color=colors)

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
# Top 10 areas with most crimes
top_areas = df['AREA NAME'].value_counts().head(10)

plt.figure(figsize=(10,5))
top_areas.plot(kind='bar')

plt.title('Top 10 Areas by Crime Count')
plt.xlabel('Area Name')
plt.ylabel('Number of Crimes')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df['Vict Age'], bins=30, edgecolor='black')

plt.title('Distribution of Victim Age')
plt.xlabel('Age')
plt.ylabel('Number of Victims')

plt.tight_layout()
plt.show()
```
